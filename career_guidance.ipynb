```python
from dotenv import load_dotenv
import os
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
google_api_key = os.getenv("GOOGLE_API_KEY")
chat = ChatGoogleGenerativeAI(api_key=google_api_key, temperature = 0.0, model = 'gemini-2.0-flash')
```

```python
from langchain.output_parsers import ResponseSchema, StructuredOutputParser
response_schema_1 = [ResponseSchema(name='traits', description='List all the traits', type='list')]
response_schema_2 = [ResponseSchema(name='career_option', description='List all the career option', type='list')]
response_schema_3 = [ResponseSchema(name='Roadmap & Pros/Cons', description='Create a dictionary of Roadmap of each career option and their pros/cons', type='dict')]
parser_1 = StructuredOutputParser.from_response_schemas(response_schema_1)
parser_2 = StructuredOutputParser.from_response_schemas(response_schema_2)
parser_3 = StructuredOutputParser.from_response_schemas(response_schema_3)
```

```python
from langchain.prompts import ChatPromptTemplate

def career_advice(user_response):
    
    prompt_1 = ChatPromptTemplate.from_template(
        """
Hey, you need to summarize the user's personality based on the user response which is : {user_response}\n
{format_instruction}
"""
    )
    chain_1 = prompt_1.partial(format_instruction = parser_1.get_format_instructions() ) | chat | parser_1
    prompt_2 = ChatPromptTemplate.from_template(
        """
Hey, you need to design a top 3 career options for the user based on the user's personality which is : {traits}\n
{format_instruction}
"""
    )
    chain_2 = prompt_2.partial(format_instruction = parser_2.get_format_instructions() ) | chat | parser_2
    prompt_3 = ChatPromptTemplate.from_template(
        """
Hey, you need to give pros/cons and 1 brief roadmap of each career option which is : {career_option}\n
{format_instruction}
"""
    )
    chain_3 = prompt_3.partial(format_instruction = parser_3.get_format_instructions() ) | chat | parser_3

    full_chain = chain_1 | chain_2 | chain_3
    result = full_chain.invoke({'user_response': user_response})
    return result
```

```python
user_response = input("Hello, what are your personalities?")
result = career_advice(user_response)
print(result)
```
